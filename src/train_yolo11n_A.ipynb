```python
from ultralytics import YOLO

model = YOLO("/content/DACS/src/yolo11n_mseaf.yaml")

model.load("yolo11n.pt")

model.info()
```

```python

model.train(
    data="VisDrone.yaml",
    epochs=100,
    patience=20,
    imgsz=640,
    batch=8,
    device=0,
    workers=2,
    seed=42
)
```
